# 03. Exploratory Inspection: Raw Parkinson's Disease Dataset

This notebook performs exploratory inspection of the raw Parkinson's Disease biomedical voice measurement dataset located at `datasets/raw/parkinsons/parkinsons.data`.

**Constraints & Scope**:
- Exploratory inspection only.
- Raw dataset is treated as read-only and is not modified.
- No cleaning, imputation, transformation, or feature scaling is performed.
- No machine-learning models are trained.
- No telemonitoring data is used.

```python
import pandas as pd
import numpy as np

# Path to the raw Parkinson's dataset
data_path = 'datasets/raw/parkinsons/parkinsons.data'
print(f"Libraries imported. Dataset path set to: {data_path}")
```

## 1. Load Dataset
Load `datasets/raw/parkinsons/parkinsons.data` using pandas.

```python
df_raw = pd.read_csv(data_path)
print(f"Dataset successfully loaded from: {data_path}")
```

## 2. Dataset Dimensions, Column Names, Inferred Types, and Sample Rows
Inspect basic dimensions, feature names, inferred data types, first 10 rows, and last 5 rows.

```python
print("=== DATASET SHAPE ===")
print(f"Rows: {df_raw.shape[0]}, Columns: {df_raw.shape[1]}")

print("\n=== COLUMN NAMES ===")
print(list(df_raw.columns))

print("\n=== PANDAS DATA TYPES ===")
print(df_raw.dtypes)

print("\n=== FIRST 10 ROWS ===")
print(df_raw.head(10).to_string())

print("\n=== LAST 5 ROWS ===")
print(df_raw.tail(5).to_string())
```

## 3. Missing Value, Empty String, and Duplicate Row Check
Check for standard pandas `NA`/`NaN` missing values, empty strings, and duplicate rows across the dataset.

```python
print("=== MISSING VALUES (NA / NaN) ===")
print(df_raw.isna().sum())

print("\n=== EMPTY STRING CHECK ===")
empty_strings = (df_raw.astype(str).apply(lambda col: col.str.strip()) == '').sum()
print(empty_strings)

print("\n=== DUPLICATE ROWS CHECK ===")
dup_rows = df_raw.duplicated().sum()
print(f"Number of completely duplicate rows: {dup_rows}")
```

## 4. Unique Values Per Feature
Inspect the number and sample of unique values for each feature.

```python
print("=== UNIQUE VALUES PER COLUMN ===")
for col in df_raw.columns:
    unique_vals = df_raw[col].unique()
    num_unique = len(unique_vals)
    print(f"\nFeature '{col}' ({num_unique} unique values):")
    if num_unique <= 10:
        print(f"  Values: {sorted(list(unique_vals))}")
    else:
        sample_vals = list(unique_vals[:5])
        print(f"  Sample values: {sample_vals} ... (total: {num_unique})")
```

## 5. Target Column (`status`) Inspection
Analyze the distribution of the target variable `status`.
Based on dataset documentation:
- `1`: Subject diagnosed with Parkinson's Disease (PD)
- `0`: Healthy control subject

```python
print("=== TARGET COLUMN 'status' INSPECTION ===")

unique_targets = df_raw['status'].unique()
print(f"Unique target values: {sorted(list(unique_targets))}")

target_counts = df_raw['status'].value_counts().sort_index()
target_pct = (df_raw['status'].value_counts(normalize=True).sort_index() * 100).round(2)

status_summary = pd.DataFrame({
    'Health Status': ['Healthy Control (0)', "Parkinson's Disease (1)"],
    'Record Count': target_counts.values,
    'Percentage (%)': target_pct.values
}, index=[0, 1])
status_summary.index.name = 'status'
print("\nTarget Distribution Table (Record Level):")
print(status_summary.to_string())
```

## 6. Numerical Descriptive Statistics
Compute summary statistics (count, mean, standard deviation, quartiles, minimum, and maximum) for all 23 numerical features.

```python
print("=== DESCRIPTIVE STATISTICS FOR NUMERICAL FEATURES ===")
num_cols = df_raw.select_dtypes(include=['float64', 'int64']).columns
stats = df_raw[num_cols].describe().T[['count', 'mean', 'std', 'min', '25%', '50%', '75%', 'max']]
print(stats.round(4).to_string())
```

## 7. Identifier and Subject Structure Analysis
Analyze the `name` column to determine whether multiple observations originate from the same subject.

Structure observed in `name`: `phon_R01_SXX_Y`, where:
- `phon_R01`: Project prefix (sustained phonations)
- `SXX`: Subject identifier (e.g., `S01`, `S02`, ..., `S50`)
- `Y`: Replicate recording index for that subject (e.g., `1`, `2`, ..., `6`, `7`)

```python
print("=== IDENTIFIER AND SUBJECT STRUCTURE ANALYSIS ===")

# Total unique names
print(f"Total row entries: {len(df_raw)}")
print(f"Total unique 'name' values: {df_raw['name'].nunique()}")

# Extract subject identifier by stripping the replicate recording suffix
df_subjects = df_raw.copy()
df_subjects['subject_id'] = df_subjects['name'].apply(lambda x: '_'.join(x.split('_')[:-1]))

unique_subjects = df_subjects['subject_id'].nunique()
print(f"\nNumber of unique individual subjects: {unique_subjects}")

# Number of records per subject
records_per_subject = df_subjects['subject_id'].value_counts()
print(f"\nDistribution of recordings per subject:")
print(records_per_subject.value_counts().rename_axis('Recordings Per Subject').reset_index(name='Number of Subjects'))

print(f"\nRecords per subject breakdown:")
for subj, cnt in records_per_subject.items():
    print(f"  {subj}: {cnt} recordings")

# Subject-level status consistency
subject_status_nunique = df_subjects.groupby('subject_id')['status'].nunique()
inconsistent_subjects = (subject_status_nunique > 1).sum()
print(f"\nSubjects with inconsistent status across recordings: {inconsistent_subjects}")

# Subject-level disease distribution
subject_level_status = df_subjects.groupby('subject_id')['status'].first()
subj_counts = subject_level_status.value_counts().sort_index()
subj_pct = (subject_level_status.value_counts(normalize=True).sort_index() * 100).round(2)

subj_summary = pd.DataFrame({
    'Health Status': ['Healthy Control (0)', "Parkinson's Disease (1)"],
    'Subject Count': subj_counts.values,
    'Percentage (%)': subj_pct.values
}, index=[0, 1])
subj_summary.index.name = 'status'
print("\nSubject-Level Distribution (Unique Individuals):")
print(subj_summary.to_string())
```

## 8. Data Integrity & Boundary Value Verification
Examine features for negative values, zeros, and boundary characteristics based strictly on the actual dataset values.

```python
print("=== DATA INTEGRITY & BOUNDARY CHECKS ===")

# Negative values check
print("Negative values check across numerical columns:")
neg_features = []
for c in num_cols:
    neg_count = (df_raw[c] < 0).sum()
    if neg_count > 0:
        neg_features.append(c)
        print(f"  Column '{c}': {neg_count} negative value(s), range [{df_raw[c].min():.4f}, {df_raw[c].max():.4f}]")

if not neg_features:
    print("  No negative values found.")
else:
    print(f"  Note: 'spread1' is an inherently negative nonlinear dynamical measure (log-based).")

# Zero values check
print("\nZero values check across numerical columns:")
for c in num_cols:
    z_cnt = (df_raw[c] == 0).sum()
    if z_cnt > 0:
        print(f"  Column '{c}': {z_cnt} zero(s)")

# Boundary checks
print("\nFrequency boundaries:")
print(f"  Fo (Average fundamental frequency): min={df_raw['MDVP:Fo(Hz)'].min():.2f} Hz, max={df_raw['MDVP:Fo(Hz)'].max():.2f} Hz")
print(f"  Fhi (Max fundamental frequency):     min={df_raw['MDVP:Fhi(Hz)'].min():.2f} Hz, max={df_raw['MDVP:Fhi(Hz)'].max():.2f} Hz")
print(f"  Flo (Min fundamental frequency):     min={df_raw['MDVP:Flo(Hz)'].min():.2f} Hz, max={df_raw['MDVP:Flo(Hz)'].max():.2f} Hz")
```

## 9. Column Descriptions and Clinical Meanings

Based on the UCI Machine Learning Repository documentation and the original study by Little et al. (2007):

1. **`name`**: ASCII recording identifier consisting of the study prefix, subject code (`S01` to `S50`), and recording trial number (`1` to `7`).
2. **Fundamental Frequency Measures**:
   - **`MDVP:Fo(Hz)`**: Average vocal fundamental frequency.
   - **`MDVP:Fhi(Hz)`**: Maximum vocal fundamental frequency.
   - **`MDVP:Flo(Hz)`**: Minimum vocal fundamental frequency.
3. **Jitter Measures (Frequency Variation / Perturbation)**:
   - **`MDVP:Jitter(%)`**: Percentage variation in fundamental frequency from cycle to cycle.
   - **`MDVP:Jitter(Abs)`**: Absolute jitter in microseconds.
   - **`MDVP:RAP`**: Relative Amplitude Perturbation.
   - **`MDVP:PPQ`**: Five-point Period Perturbation Quotient.
   - **`Jitter:DDP`**: Average absolute difference of differences between consecutive periods (DDP = 3 * RAP).
4. **Shimmer Measures (Amplitude Variation / Perturbation)**:
   - **`MDVP:Shimmer`**: Local shimmer variation in amplitude from cycle to cycle.
   - **`MDVP:Shimmer(dB)`**: Local shimmer in decibels.
   - **`Shimmer:APQ3`**: Three-point Amplitude Perturbation Quotient.
   - **`Shimmer:APQ5`**: Five-point Amplitude Perturbation Quotient.
   - **`MDVP:APQ`**: Eleven-point Amplitude Perturbation Quotient.
   - **`Shimmer:DDA`**: Average absolute difference between consecutive amplitude differences (DDA = 3 * APQ3).
5. **Harmonicity and Noise Measures**:
   - **`NHR`**: Noise-to-Harmonics Ratio (ratio of acoustic noise to tonal voice component).
   - **`HNR`**: Harmonics-to-Noise Ratio (dB). Lower HNR indicates more breathiness and dysphonia.
6. **Target Variable**:
   - **`status`**: Health status: `1` = Parkinson's Disease (PD), `0` = Healthy control.
7. **Nonlinear Dynamical Complexity Measures**:
   - **`RPDE`**: Recurrence Period Density Entropy (vocal tract periodicity measure).
   - **`DFA`**: Detrended Fluctuation Analysis (signal fractal scaling exponent).
   - **`spread1`**: Nonlinear measure of fundamental frequency variation (negative log-scaled).
   - **`spread2`**: Nonlinear measure of fundamental frequency variation (spectral spread).
   - **`D2`**: Correlation dimension (measure of complexity / chaotic attractor dimension).
   - **`PPE`**: Pitch Period Entropy (measure of impaired vocal control).

- **Critical Implications for Avoiding Data Leakage in Model Evaluation**:
  - Because each of the 32 subjects contributed multiple sustained vowel phonations, individual voice biometrics (anatomy, accent, fundamental vocal tone) are shared across all recordings of that subject.
  - **Standard random train/test splitting or random K-fold cross-validation would cause severe data leakage**, placing recordings of the same patient into both the training and testing sets. This would lead to models that identify individual speakers rather than generalizable Parkinsonian dysphonia patterns.
  - Future model development and cross-validation MUST use **Subject-Aware Group Splitting** (e.g., `GroupKFold` or `StratifiedGroupKFold` grouped by `subject_id`) to ensure no subject appears in both train and validation/test folds.
