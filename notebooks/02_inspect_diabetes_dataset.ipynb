# 02. Exploratory Inspection: Raw Diabetes Dataset

This notebook performs exploratory data inspection on the raw Pima Indians Diabetes dataset located at `datasets/raw/diabetes/diabetes.csv`.

**Constraints & Guidelines**:
- Exploratory inspection only.
- Raw CSV is treated as read-only and is not modified.
- No data cleaning, imputation, transformation, or feature replacement is performed at this stage.
- No machine-learning models are trained.

```python
import pandas as pd
import numpy as np

# Path to the raw diabetes dataset
data_path = 'datasets/raw/diabetes/diabetes.csv'
print(f"Libraries imported. Dataset path set to: {data_path}")
```

## 1. Load Dataset
Load `datasets/raw/diabetes/diabetes.csv` using pandas.

```python
df_raw = pd.read_csv(data_path)
print(f"Dataset successfully loaded from: {data_path}")
```

## 2. Dataset Dimensions, Feature Types, and Sample Rows
Inspect basic dimensions, column names, inferred pandas data types, the first 10 rows, and the last 5 rows.

```python
print("=== DATASET SHAPE ===")
print(f"Rows: {df_raw.shape[0]}, Columns: {df_raw.shape[1]}")

print("\n=== COLUMN NAMES ===")
print(list(df_raw.columns))

print("\n=== PANDAS DATA TYPES ===")
print(df_raw.dtypes)

print("\n=== FIRST 10 ROWS ===")
print(df_raw.head(10).to_string())

print("\n=== LAST 5 ROWS ===")
print(df_raw.tail(5).to_string())
```

## 3. Missing Value and Empty String Inspection
Check for standard pandas `NA` / `NaN` nulls and examine if empty or whitespace-only strings exist.

```python
print("=== STANDARD MISSING VALUES (NA / NaN) ===")
print(df_raw.isna().sum())

print("\n=== EMPTY STRING CHECK ===")
empty_strings = (df_raw.astype(str).apply(lambda col: col.str.strip()) == '').sum()
print(empty_strings)
```

## 4. Unique Values Per Feature
Inspect the distinct values present in every column to evaluate continuous vs. discrete distributions.

```python
print("=== UNIQUE VALUES PER COLUMN ===")
for col in df_raw.columns:
    unique_vals = df_raw[col].unique()
    num_unique = len(unique_vals)
    print(f"\nColumn '{col}' ({num_unique} unique values):")
    if num_unique <= 15:
        sorted_vals = sorted(list(unique_vals))
        print(f"  Values: {sorted_vals}")
    else:
        sample_vals = list(unique_vals[:8])
        print(f"  Sample values: {sample_vals} ... (total: {num_unique})")
```

## 5. Target Column (`Outcome`) Distribution
Inspect unique values, counts, and class balance percentages for the target variable `Outcome`.

```python
print("=== TARGET COLUMN 'Outcome' INSPECTION ===")

unique_outcomes = df_raw['Outcome'].unique()
print(f"Unique target values: {sorted(list(unique_outcomes))}")

counts = df_raw['Outcome'].value_counts().sort_index()
percentages = (df_raw['Outcome'].value_counts(normalize=True).sort_index() * 100).round(2)

outcome_summary = pd.DataFrame({
    'Diagnosis': ['Non-Diabetic (0)', 'Diabetic (1)'],
    'Count': counts.values,
    'Percentage (%)': percentages.values
}, index=[0, 1])
outcome_summary.index.name = 'Outcome'
print("\nTarget Distribution Table:")
print(outcome_summary.to_string())
```

## 6. Duplicate Rows Check
Check whether the raw dataset contains any duplicate rows.

```python
print("=== DUPLICATE ROWS CHECK ===")
duplicate_count = df_raw.duplicated().sum()
print(f"Number of completely duplicate rows: {duplicate_count}")
```

## 7. Descriptive Statistics
Summary statistics (mean, standard deviation, quartiles, min/max) for all numerical columns.

```python
print("=== DESCRIPTIVE STATISTICS FOR NUMERICAL COLUMNS ===")
stats = df_raw.describe().T[['count', 'mean', 'std', 'min', '25%', '50%', '75%', 'max']]
print(stats.round(2).to_string())
```

## 8. Specific Zero-Value Audit for Medical Features
Audit five physiological features (`Glucose`, `BloodPressure`, `SkinThickness`, `Insulin`, `BMI`) where a value of zero is biologically implausible in living individuals.

*Important Note: We only identify and quantify zero values here; no replacements or imputations are performed.*

```python
print("=== ZERO-VALUE AUDIT IN PHYSIOLOGICALLY CRITICAL FEATURES ===")
med_features = ['Glucose', 'BloodPressure', 'SkinThickness', 'Insulin', 'BMI']

zero_audit = []
for feat in med_features:
    zero_cnt = (df_raw[feat] == 0).sum()
    zero_pct = (zero_cnt / len(df_raw)) * 100
    zero_audit.append({
        'Feature': feat,
        'Zero Count': zero_cnt,
        'Percentage (%)': round(zero_pct, 2),
        'Plausibility': 'Biologically impossible in living humans (hidden missing data)'
    })

zero_df = pd.DataFrame(zero_audit)
print(zero_df.to_string(index=False))
```

## 9. Suspicious and Extreme Value Verification
Examine boundary values across all features based strictly on observations from the actual dataset.

```python
print("=== DATA INTEGRITY & SUSPICIOUS VALUE CHECK ===")

# 1. Negative values check
negatives = (df_raw < 0).sum()
has_negatives = (negatives > 0).any()
print(f"Negative values across all columns: {'Found' if has_negatives else 'None (all values >= 0)'}")

# 2. Extreme upper-boundary checks observed in actual data
print("\nNotable boundary values observed in the actual dataset:")
print(f"- Max SkinThickness: {df_raw['SkinThickness'].max()} mm (Observations > 60 mm: {(df_raw['SkinThickness'] > 60).sum()})")
print(f"- Max Insulin: {df_raw['Insulin'].max()} mu U/ml (Observations > 600 mu U/ml: {(df_raw['Insulin'] > 600).sum()})")
print(f"- Max Pregnancies: {df_raw['Pregnancies'].max()} (Observations >= 13: {(df_raw['Pregnancies'] >= 13).sum()})")
print(f"- Max BMI: {df_raw['BMI'].max()} kg/m^2 (Observations >= 50: {(df_raw['BMI'] >= 50).sum()})")
print(f"- Age Range: min={df_raw['Age'].min()}, max={df_raw['Age'].max()} years")
```

## 10. Column Descriptions and Clinical Meanings

Based on official documentation for the National Institute of Diabetes and Digestive and Kidney Diseases (NIDDK) / Pima Indians Diabetes Dataset:

1. **`Pregnancies`**: Number of times the patient has been pregnant (discrete integer, range: 0 to 17).
2. **`Glucose`**: Plasma glucose concentration at 2 hours in an oral glucose tolerance test (continuous integer, mg/dl). A value of 0 is physiologically invalid.
3. **`BloodPressure`**: Diastolic blood pressure (continuous integer, mm Hg). A value of 0 indicates absence of measurement rather than true zero pressure.
4. **`SkinThickness`**: Triceps skin fold thickness (continuous integer, mm). Used to estimate body fat percentage. Values of 0 indicate missing measurement.
5. **`Insulin`**: 2-Hour serum insulin (continuous integer, mu U/ml). Values of 0 represent unmeasured or missing data.
6. **`BMI`**: Body Mass Index (continuous float, weight in kg / (height in m)^2). A value of 0.0 is biologically impossible for a living person.
7. **`DiabetesPedigreeFunction`**: A genetic scoring metric calculated based on family history of diabetes (continuous float, range: 0.078 to 2.42).
8. **`Age`**: Age in years (discrete integer, range: 21 to 81). All subjects in this cohort are adult females at least 21 years of age.
9. **`Outcome`**: Target binary variable:
   - `0`: Non-diabetic (negative test result within 5 years)
   - `1`: Diabetic (positive test result within 5 years)
