# 01. Exploratory Inspection: Raw Heart Disease Dataset

This notebook performs an initial exploratory inspection of the raw Heart Disease dataset (`processed.cleveland.data`).

**Constraints & Scope**:
- Exploratory inspection only.
- Raw dataset is treated as read-only and is not modified.
- No cleaning, imputation, transformation, or ML model training is performed.


```python
import pandas as pd
import numpy as np

# Official standard 14-column naming
columns = [
    'age',
    'sex',
    'cp',
    'trestbps',
    'chol',
    'fbs',
    'restecg',
    'thalach',
    'exang',
    'oldpeak',
    'slope',
    'ca',
    'thal',
    'num'
]

# Relative path from project root
data_path = 'datasets/raw/heart/processed.cleveland.data'
print("Configuration and imports initialized.")
```

## 1. Load Dataset Without Header Assumptions
The raw data file does not include header row labels. We load it using `header=None` and assign the official 14 standard feature names.


```python
# Load dataset without assuming existing headers in the file
df_raw = pd.read_csv(data_path, header=None, names=columns)
print(f"Successfully loaded raw dataset from: {data_path}")
```

## 2. Dataset Shape, Sample Rows, and Data Types
Inspect basic dimensions, sample observations, and inferred pandas data types.


```python
print("=== DATASET SHAPE ===")
print(f"Rows: {df_raw.shape[0]}, Columns: {df_raw.shape[1]}")

print("\n=== COLUMN NAMES ===")
print(list(df_raw.columns))

print("\n=== PANDAS DATA TYPES ===")
print(df_raw.dtypes)

print("\n=== FIRST 10 ROWS ===")
print(df_raw.head(10).to_string())

print("\n=== LAST 5 ROWS ===")
print(df_raw.tail(5).to_string())
```

## 3. Missing Value and Non-Standard Symbol Inspection
Inspect missing values:
- Check standard pandas `NA` / `NaN` counts.
- Specifically identify `'?'` sentinel values before any type conversion.
- Identify which columns contain `'?'` values.


```python
print("=== STANDARD MISSING VALUES (NA / NaN) ===")
print(df_raw.isna().sum())

print("\n=== IDENTIFYING '?' VALUES ===")
q_mark_counts = (df_raw == '?').sum()
print("Count of '?' per column:")
print(q_mark_counts)

print("\n=== COLUMNS CONTAINING '?' VALUES ===")
cols_with_q = q_mark_counts[q_mark_counts > 0]
for col, cnt in cols_with_q.items():
    pct = (cnt / len(df_raw)) * 100
    print(f"Column '{col}': {cnt} occurrence(s) of '?' ({pct:.2f}%)")
```

## 4. Unique Values Per Feature
Inspect the distinct values present in every column to understand continuous vs. categorical feature distributions.


```python
print("=== UNIQUE VALUES PER FEATURE ===")
for col in df_raw.columns:
    unique_vals = df_raw[col].unique()
    print(f"\nFeature '{col}' ({len(unique_vals)} unique values):")
    if len(unique_vals) <= 15:
        sorted_vals = sorted(list(unique_vals), key=lambda x: str(x))
        print(f"  Values: {sorted_vals}")
    else:
        sample_vals = list(unique_vals[:8])
        print(f"  Sample values: {sample_vals} ... (total: {len(unique_vals)})")
```

## 5. Target Column (`num`) Distribution
Inspect the distribution of the target variable `num` in both multiclass (0 to 4) and clinical binary format (0 vs > 0).


```python
print("=== TARGET COLUMN 'num' INSPECTION ===")

target_counts = df_raw['num'].value_counts().sort_index()
target_pct = (df_raw['num'].value_counts(normalize=True).sort_index() * 100).round(2)

target_summary = pd.DataFrame({
    'Count': target_counts,
    'Percentage (%)': target_pct
})
target_summary.index.name = 'num'
print(target_summary)

print("\n=== BINARY REGROUPING (0 = Absence, 1-4 = Presence) ===")
has_disease = (df_raw['num'] > 0).astype(int)
binary_counts = has_disease.value_counts().sort_index()
binary_pct = (has_disease.value_counts(normalize=True).sort_index() * 100).round(2)
binary_summary = pd.DataFrame({
    'Diagnosis': ['Absence (0)', 'Presence (1-4)'],
    'Count': binary_counts.values,
    'Percentage (%)': binary_pct.values
})
print(binary_summary.to_string(index=False))
```

## 6. Duplicate Rows Check
Check whether the raw dataset contains any duplicate rows.


```python
print("=== DUPLICATE ROWS CHECK ===")
dup_count = df_raw.duplicated().sum()
print(f"Number of duplicate rows: {dup_count}")
```

## 7. Basic Descriptive Statistics
Summary statistics (mean, standard deviation, quartiles, min/max) for all numerical features. A temporary numeric view is used for `ca` and `thal` purely for reporting purposes without altering `df_raw`.


```python
print("=== DESCRIPTIVE STATISTICS FOR NUMERICAL COLUMNS ===")
# Note: 'ca' and 'thal' were inferred as object dtype because of '?' strings.
# We create a temporary numeric view strictly for displaying descriptive stats without modifying df_raw.
df_numeric_view = df_raw.copy()
df_numeric_view['ca'] = pd.to_numeric(df_numeric_view['ca'].replace('?', np.nan))
df_numeric_view['thal'] = pd.to_numeric(df_numeric_view['thal'].replace('?', np.nan))

stats = df_numeric_view.describe().T[['count', 'mean', 'std', 'min', '25%', '50%', '75%', 'max']]
print(stats.round(2).to_string())
```

## 8. Feature Explanations and Data Quality Observations

### 8.1 Representation of the 14 Columns
1. **`age`**: Age of the patient in years (numeric, range: 29.0 to 77.0).
2. **`sex`**: Biological sex of the patient (nominal binary: `1.0` = male, `0.0` = female).
3. **`cp`**: Chest pain type (nominal categorical, 4 categories):
   - `1.0`: Typical angina
   - `2.0`: Atypical angina
   - `3.0`: Non-anginal pain
   - `4.0`: Asymptomatic
4. **`trestbps`**: Resting blood pressure in mm Hg on admission to hospital (numeric, range: 94.0 to 200.0).
5. **`chol`**: Serum cholesterol in mg/dl (numeric, range: 126.0 to 564.0).
6. **`fbs`**: Fasting blood sugar > 120 mg/dl (nominal binary: `1.0` = true, `0.0` = false).
7. **`restecg`**: Resting electrocardiographic results (nominal categorical, 3 categories):
   - `0.0`: Normal
   - `1.0`: ST-T wave abnormality
   - `2.0`: Probable or definite left ventricular hypertrophy by Estes' criteria
8. **`thalach`**: Maximum heart rate achieved during exercise testing (numeric, range: 71.0 to 202.0).
9. **`exang`**: Exercise-induced angina (nominal binary: `1.0` = yes, `0.0` = no).
10. **`oldpeak`**: ST depression induced by exercise relative to rest (numeric continuous, range: 0.0 to 6.2).
11. **`slope`**: The slope of the peak exercise ST segment (ordinal categorical, 3 categories):
    - `1.0`: Upsloping
    - `2.0`: Flat
    - `3.0`: Downsloping
12. **`ca`**: Number of major vessels (0–3) colored by fluoroscopy (numeric discrete integer represented as strings due to `'?'`).
13. **`thal`**: Thallium scintigraphy / cardiac defect (nominal categorical):
    - `3.0`: Normal
    - `6.0`: Fixed defect
    - `7.0`: Reversible defect
    - `'?'`: Missing observation
14. **`num`**: Diagnosis of heart disease (angiographic disease status; **target variable**).

### 8.2 Target Variable Analysis
- **Target column**: `num`
- **Current target format**: Integer values ranging from `0` to `4`:
  - `0`: Absence of heart disease (< 50% diameter narrowing). Total: 164 instances (54.13%).
  - `1`: Stage 1 disease. Total: 55 instances (18.15%).
  - `2`: Stage 2 disease. Total: 36 instances (11.88%).
  - `3`: Stage 3 disease. Total: 35 instances (11.55%).
  - `4`: Stage 4 disease. Total: 13 instances (4.29%).
- When formulated as a binary classification task (0 = absence vs. 1–4 = presence), the distribution is:
  - Absence (0): 164 samples (54.13%)
  - Presence (1–4): 139 samples (45.87%)

### 8.3 Missing Values and Encoding Issues
- Standard pandas `isna().sum()` returns **0** for all columns because missing values were encoded as the string literal `'?'` instead of standard IEEE NaN or nulls.
- Columns containing `'?'`:
  - `ca`: 4 missing values (1.32% of total rows).
  - `thal`: 2 missing values (0.66% of total rows).
- Due to the presence of `'?'`, pandas loaded `ca` and `thal` as generic `object` dtype rather than numeric.

### 8.4 Data Quality Concerns Discovered
1. **Sentinel Strings as Missing Data**: The `'?'` string values must be handled during future preprocessing.
2. **Object Data Types**: `ca` and `thal` must be converted to numeric representation once `'?'` is resolved.
3. **Categorical Features Encoded as Floats**: Features such as `sex`, `cp`, `fbs`, `restecg`, `exang`, `slope`, and `thal` are stored as floating-point numbers (`float64`) in the raw file despite representing discrete categories.
4. **Extreme / Outlier Values**: In `chol`, one patient has a serum cholesterol level of 564.0 mg/dl, requiring consideration during scaling or outlier analysis.
5. **Multiclass vs. Binary Target**: The raw target has 5 classes (0 through 4), with class 4 containing only 13 samples (4.29%), which would cause severe imbalance if used in multiclass prediction without appropriate strategy.
